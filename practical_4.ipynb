```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Generate data from a normal distribution
# Using np.random.randn to generate samples from a standard normal distribution (mean=0, std=1)
num_samples = 1000
data = np.random.randn(num_samples)

# 2. Plot the histogram
plt.figure(figsize=(10, 6))
sns.histplot(data, bins=30, kde=True, color='skyblue')

# Add title and labels
plt.title('Histogram of a Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)

# Display the plot
plt.show()
```

The code above first generates 1000 random numbers from a standard normal distribution (mean=0, standard deviation=1) using `np.random.randn`. Then, it uses `seaborn.histplot` to create a histogram of this data, including a Kernel Density Estimate (KDE) line to show the shape of the distribution. Finally, it adds a title, axis labels, and a grid for better readability before displaying the plot.

```python
# Calculate descriptive statistics
mean_val = np.mean(data)
median_val = np.median(data)
variance_val = np.var(data)
std_dev_val = np.std(data)

print(f"Mean: {mean_val:.4f}")
print(f"Median: {median_val:.4f}")
print(f"Variance: {variance_val:.4f}")
print(f"Standard Deviation: {std_dev_val:.4f}")
```

This code block calculates the mean, median, variance, and standard deviation of the `data` array (which contains 1000 samples from a standard normal distribution) using `numpy` functions and then prints these values. For a standard normal distribution, the theoretical mean is 0 and the standard deviation is 1, so the computed values should be close to these.

```python
import numpy as np

# Simulate a coin toss experiment
# Let's define 0 as tails and 1 as heads
num_tosses = 1000
toss_results = np.random.randint(0, 2, num_tosses)

# Count heads and tails
heads_count = np.sum(toss_results == 1)
tails_count = np.sum(toss_results == 0)

# Compute probabilities
probability_heads = heads_count / num_tosses
probability_tails = tails_count / num_tosses

print(f"Total tosses: {num_tosses}")
print(f"Heads count: {heads_count}")
print(f"Tails count: {tails_count}")
print(f"Probability of Heads: {probability_heads:.4f}")
print(f"Probability of Tails: {probability_tails:.4f}")
```

This code simulates a coin toss experiment. It uses `np.random.randint(0, 2, num_tosses)` to generate `num_tosses` (1000 in this case) random integers, where 0 represents tails and 1 represents heads. It then counts the occurrences of heads and tails and calculates their respective probabilities by dividing the count by the total number of tosses. Due to randomness, these probabilities will be close to 0.5 but might not be exactly 0.5.
